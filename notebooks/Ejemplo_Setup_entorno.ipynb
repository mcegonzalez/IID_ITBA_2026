# Setup del repositorio

```python
import os
import sys
from google.colab import drive, userdata

# Acceder a nuestro Google Drive
drive.mount('/content/drive')

# Definir direcciones de trabajo
DRIVE_PATH = '/content/drive/MyDrive/ITBA'
REPO_NAME = 'Curso_Introduccion_Ingenieria_Datos'
PROJECT_PATH = os.path.join(DRIVE_PATH, REPO_NAME)

# Obtener credenciales
pat = userdata.get('GITHUB_PAT')
username = userdata.get('USERNAME')

#Ir a carpeta raíz en Google Drive
%cd {DRIVE_PATH}

try:
    if not os.path.exists(PROJECT_PATH):
      #Clonar repositorio github en nuestro drive
      !git clone https://{username}:{pat}@github.com/{username}/{REPO_NAME}.git

except Exception as e:
    print(f"Fallo clonado del repo git. Error: {e}")

# Agregamos el directorio 'src' para que Python pueda importar los archivos como librerias.
# Ejemplo para ejecutar la función Inferencia desde el archivo modelos.py en carpeta src: from src.modelos import Inferencia
sys.path.append(os.path.join(PROJECT_PATH, 'src'))
```

## Crear estructura

```python
#Creamos las carpetas y un archivo de ejemplo para que Git lo registre.
os.makedirs(PROJECT_PATH+"/data", exist_ok=True)
!touch {PROJECT_PATH}/data/.gitkeep
os.makedirs(PROJECT_PATH+"/data/raw", exist_ok=True)
!touch {PROJECT_PATH}/data/raw/.gitkeep
os.makedirs(PROJECT_PATH+"/data/processed", exist_ok=True)
!touch {PROJECT_PATH}/data/processed/.gitkeep
os.makedirs(PROJECT_PATH+"/notebooks", exist_ok=True)
!touch {PROJECT_PATH}/notebooks/.gitkeep
os.makedirs(PROJECT_PATH+"/src", exist_ok=True)
!touch {PROJECT_PATH}/src/.gitkeep
```

```python
#Actualizar repositorio con estructura
%cd {PROJECT_PATH}

# Configurar credenciales para git
!git config --global user.email {userdata.get('USEREMAIL')}
!git config --global user.name {userdata.get('GITUSERNAME')}

!git add *
!git commit -m "Creación de estructura"
!git push
```
